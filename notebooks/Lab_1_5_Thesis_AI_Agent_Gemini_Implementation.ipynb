```python
import google.generativeai as genai

genai.configure(api_key=GEMINI_KEY)
model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
import requests

zero_shot_url = "https://raw.githubusercontent.com/MirmatlabKarimli/Thesis_homework/main/Homework%204/Prompt%201%20-%20Zeroshot.txt"
few_shot_url  = "https://raw.githubusercontent.com/MirmatlabKarimli/Thesis_homework/main/Homework%204/Prompt%202%20-%20Fewshot.txt"

zero_shot_prompt = requests.get(zero_shot_url).text
few_shot_prompt  = requests.get(few_shot_url).text

print("Zero-shot Prompt Loaded:\n", zero_shot_prompt[:300], "...\n")
print("Few-shot Prompt Loaded:\n", few_shot_prompt[:300], "...\n")
```

```python
response_zero = model.generate_content(zero_shot_prompt)

print("=== ZERO-SHOT PROMPT EXECUTION ===")
print(response_zero.text)
```

```python
response_few = model.generate_content(few_shot_prompt)

print("=== FEW-SHOT PROMPT EXECUTION ===")
print(response_few.text)
```

```python
sample_data = """
Drug: Paracetamol
Previous 7-day sales: 120, 150, 160, 140, 180, 200, 220
Season: Winter
Price: 5.20
Promotion: No
"""
print(sample_data)
```

```python
reasoning_prompt = f"""
You are an AI agent designed for drug sales forecasting as described in the thesis.
Follow the internal pipeline:

1. Data Understanding
2. Identify seasonal & historical patterns
3. Infer demand trend
4. Generate predicted sales output

Input data:
{sample_data}

Explain your reasoning step-by-step, then give the final forecast number.
"""

response_reasoning = model.generate_content(reasoning_prompt)
print(response_reasoning.text)
```

```python
final_prompt = f"""
Provide ONLY the final predicted sales for Paracetamol next week,
based on the previous reasoning and the sample dataset.
Input:
{sample_data}
"""

response_output = model.generate_content(final_prompt)
print("Predicted Sales Output:", response_output.text)
```

```python
print("""
AI Forecasting Pipeline
---------------------------
[1] Input Data
      ↓
[2] Data Understanding
      ↓
[3] Reasoning & Pattern Detection
      ↓
[4] Sales Prediction Output
---------------------------
""")
```
